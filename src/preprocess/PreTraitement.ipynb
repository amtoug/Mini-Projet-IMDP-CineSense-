```python
#from datasets import load_dataset
import pandas as pd
from preprocess import PreTraitement,generer_donnees,generer_et_traiter_donnees
from sklearn.model_selection import train_test_split
from Vectorisation import Vectorisation

```

```python
df=generer_et_traiter_donnees("../../data/data_clean.csv")
```

```python
df.head()
df.columns
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df["review"], df["sentiment"],
    test_size=0.2,
    random_state=42,
    stratify=df["sentiment"] #Avec stratify=y =>train : ~70 % / 30 %test : ~70 % / 30
)
```

```python
vec = Vectorisation()

X_train_vec = vec.fit_transform(X_train)
X_test_vec  = vec.transform(X_test)
```

```python
feature_names = vec.vectorizer.get_feature_names_out()

X_train_df = pd.DataFrame(
    X_train_vec.toarray(),
    columns=feature_names
)

X_test_df = pd.DataFrame(
    X_test_vec.toarray(),
    columns=feature_names
)
```

```python
X_train_df["label"] = y_train.values
X_test_df["label"] = y_test.values
```

```python
X_train_df.to_csv("../../data/train_clean.csv", index=False)
X_test_df.to_csv("../../data/test_clean.csv", index=False)
```
